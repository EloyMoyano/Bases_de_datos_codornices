```python
from pathlib import Path                          

import pandas as pd
import matplotlib.pyplot as plt
```

```python
base = (
    Path.home()
    / "Trabajo_Especial"
    / "Bases_de_datos_codornices"
    / "data_raw"
    / "T01"
)

archivo_acc = base / "T01_sincomportamiento.csv"
archivo_comp = base / "T01_comportamientos.csv"

print(archivo_acc)
print(archivo_comp)
```

```python
acelerometria = pd.read_csv(archivo_acc, index_col=0)
comportamientos = pd.read_csv(archivo_comp, index_col=0)

print("Acelerometría:", acelerometria.shape)
print("Comportamientos:", comportamientos.shape)
```

```python
import matplotlib.pyplot as plt

# Parámetros
fs = 25
inicio = 6275
fin_etiqueta = 6299

duracion_segundos = 30
fin_grafico = inicio + fs * duracion_segundos

# Segmento desde el comienzo etiquetado
segmento = datos.loc[
    inicio:min(fin_grafico, datos.index.max())
].copy()

# Tiempo en segundos desde el inicio del baño
segmento["tiempo_relativo"] = (
    segmento.index - inicio
) / fs

# Duración del fragmento etiquetado
fin_etiqueta_relativo = (
    fin_etiqueta - inicio
) / fs

# Gráficos
fig, ejes = plt.subplots(
    nrows=3,
    ncols=1,
    figsize=(14, 10),
    sharex=True
)

columnas = ["Axis-X", "Axis-Y", "Axis-Z"]
colores = ["tab:blue", "tab:orange", "tab:green"]

for eje, columna, color in zip(ejes, columnas, colores):

    eje.plot(
        segmento["tiempo_relativo"],
        segmento[columna],
        color=color,
        linewidth=1
    )

    # Intervalo originalmente etiquetado
    eje.axvspan(
        0,
        fin_etiqueta_relativo,
        color="red",
        alpha=0.25,
        label="Etiqueta inicial"
    )

    eje.set_ylabel(columna)
    eje.set_xlim(0, duracion_segundos)
    eje.grid(alpha=0.3)
    eje.legend(loc="upper right")

ejes[-1].set_xlabel("Tiempo desde la etiqueta (s)")

fig.suptitle(
    "Aceleración durante los 30 segundos posteriores a la etiqueta",
    fontsize=14
)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

fs = 25

# Segundo segmento etiquetado
inicio = 43600
fin_etiqueta = 43649

duracion_segundos = 30
fin_grafico = inicio + fs * duracion_segundos

segmento = datos.loc[
    inicio:min(fin_grafico, datos.index.max())
].copy()

segmento["tiempo_relativo"] = (
    segmento.index - inicio
) / fs

fin_etiqueta_relativo = (
    fin_etiqueta - inicio
) / fs

fig, ejes = plt.subplots(
    3,
    1,
    figsize=(14, 10),
    sharex=True
)

columnas = ["Axis-X", "Axis-Y", "Axis-Z"]
colores = ["tab:blue", "tab:orange", "tab:green"]

for eje, columna, color in zip(ejes, columnas, colores):

    eje.plot(
        segmento["tiempo_relativo"],
        segmento[columna],
        color=color,
        linewidth=1
    )

    eje.axvspan(
        0,
        fin_etiqueta_relativo,
        color="red",
        alpha=0.25,
        label="Etiqueta de baño"
    )

    eje.set_ylabel(columna)
    eje.set_xlim(0, duracion_segundos)
    eje.grid(alpha=0.3)
    eje.legend(loc="upper right")

ejes[-1].set_xlabel("Tiempo desde el inicio de la etiqueta (s)")

fig.suptitle(
    "Segundo segmento etiquetado como baño de arena",
    fontsize=14
)

plt.tight_layout()
plt.show()
```
